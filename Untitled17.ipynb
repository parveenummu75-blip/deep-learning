```python
import sympy as sp

# Define variable
x = sp.Symbol('x')

# Function
f = x**2

# Gradient
gradient = sp.diff(f, x)

# Given values
wt = 0.5
mu = 0.1
vt = 2
eta = 0.5

# Step i: Look Ahead Point
lap = wt + mu * vt

# Step ii: Gradient at LAP
g = gradient.subs(x, lap)

# Step iii: Update Velocity
vt1 = mu * vt - eta * g

# Step iv: Update Weight
wt1 = wt + vt1

print("LAP =", lap)
print("Gradient =", g)
print("V(t+1) =", vt1)
print("W(t+1) =", wt1)
```
